# 🎓 Análise de Dados do ProUni
Este notebook realiza a análise da base de dados de cursos, mensalidades e bolsas do ProUni.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Carregar a base de dados
# No Colab, você precisará fazer o upload do arquivo ou conectar ao GitHub
try:
    df = pd.read_csv('data/2018-02-08/cursos-prouni.csv')
    print('Base carregada com sucesso!')
except FileNotFoundError:
    print('Erro: Arquivo não encontrado. Certifique-se de que o arquivo cursos-prouni.csv está na pasta correta.')
```

## 🛠️ Limpeza e Preparação
Ajustando formatos de números e tratando valores ausentes.

```python
# Preencher valores nulos de bolsas com 0
cols_bolsas = ['bolsa_integral_cotas', 'bolsa_integral_ampla', 'bolsa_parcial_cotas', 'bolsa_parcial_ampla']
df[cols_bolsas] = df[cols_bolsas].fillna(0).astype(int)

# Calcular total de bolsas por curso
df['total_bolsas'] = df[cols_bolsas].sum(axis=1)

df.head()
```

## 📊 Análises Principais
1. Cursos com as maiores mensalidades
2. Distribuição de bolsas por curso

```python
# Top 10 cursos mais caros
top_caros = df.nlargest(10, 'mensalidade')[['nome', 'universidade_nome', 'mensalidade']]
print('--- Top 10 Cursos Mais Caros ---')
print(top_caros)

# Gráfico de Mensalidades dos Top 10
plt.figure(figsize=(12,6))
sns.barplot(data=top_caros, x='mensalidade', y='nome', hue='universidade_nome', dodge=False)
plt.title('Top 10 Cursos com Maiores Mensalidades')
plt.xlabel('Mensalidade (R$)')
plt.ylabel('Curso')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

## 📈 Análise de Bolsas
Quais cursos oferecem a maior quantidade de bolsas no total?

```python
# Top 10 cursos com mais bolsas
top_bolsas = df.nlargest(10, 'total_bolsas')[['nome', 'total_bolsas', 'universidade_nome']]
print('--- Top 10 Cursos com Mais Bolsas ---')
print(top_bolsas)

plt.figure(figsize=(12,6))
sns.barplot(data=top_bolsas, x='total_bolsas', y='nome', color='skyblue')
plt.title('Top 10 Cursos com Maior Quantidade de Bolsas')
plt.xlabel('Total de Bolsas')
plt.ylabel('Curso')
plt.show()
```
